```python
import requests as req
import pandas as pd
import numpy as np


url="https://dummyjson.com/products"

response=req.get(url)
print(response)
data=response.json()
print(response.status_code)
df=pd.DataFrame(data)
df
```

```python
df.columns
```

```python
df=pd.DataFrame(data["products"])
df
```

```python
df["products"][0]["price"]
```

```python
df.columns
```

```python
df=df[["price","discountPercentage","rating"]]
df
```

```python
df.isna().sum()
```

```python
df.dtypes
```

```python

Y=df.iloc[:,0]
Y
df=df.drop(columns=["price"])
X=df
X
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3)
```

```python
from sklearn.preprocessing import StandardScaler

sc=StandardScaler()

X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)

```

```python
from sklearn.linear_model import LinearRegression

lr=LinearRegression()

lr=lr.fit(X_train,Y_train)

Y_pred=lr.predict(X_test)

Y_pred
```

```python
from sklearn.metrics import r2_score,mean_absolute_percentage_error

print(f"R2_Score = {r2_score(Y_test,Y_pred)}")
print(f"mean_absolute_percentage_error = {mean_absolute_percentage_error(Y_test,Y_pred)}")
```

```python
from sklearn.preprocessing import PolynomialFeatures

pf=PolynomialFeatures(degree=4)

X_train_poly=pf.fit_transform(X_train)
X_test_poly=pf.transform(X_test)

lr_2=LinearRegression()

lr_2=lr_2.fit(X_train_poly,Y_train)

Y_pred_2=lr_2.predict(X_test_poly)
```

```python
print(f"R2_Score = {r2_score(Y_test,Y_pred_2)}")
print(f"mean_absolute_percentage_error = {mean_absolute_percentage_error(Y_test,Y_pred_2)}")
```
