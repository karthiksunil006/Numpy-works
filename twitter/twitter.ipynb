```python
import pandas as pd
import numpy as np
import re # This module provides pattern matching operations using list of strings given
import nltk
```

```python
data=pd.read_csv("twitter_validation.csv",header=None)
df=pd.DataFrame(data)
df
```

```python
df.columns=["ID","Platform","Nature","Comment"]
df
```

```python
df["Comment"]=df["Comment"].str.lower()
df["Comment"]
```

```python
def replace_text(text):

    text=re.sub(r"http\S+","",text) # replace all the urls starting with http or https with empty string
    # \S+ matches any non whitespace characters

    text=re.sub(r"[^a-z\s]","",text) # replace all elements 
    # \s matches a single whitespace character

    return text

df["Comment"]=df["Comment"].apply(replace_text)
```
