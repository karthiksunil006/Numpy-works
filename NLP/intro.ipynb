NLP - Natural Language Processing
---------------------------------

> Feild of Artificial Intelligence (AI) that enables computers to understand , interpret , and generate human language
> which convert our language into machine knowing language (text message to machine knowing)

Preprocessing Steps
-------------------

Step-1 > Remove unwanted white spaces
Step-2 > convert the characters into ASCII codes
Step-3 > Expand the shortforms
Step-4 > Remove the unwanted special characters like (%,{},.....etc.)


Must Implement Steps
--------------------

" Universities are amazing places for studying and learning new things "

> Remove the stop words
    eg:are,for,and,the,is.....etc.

> Convert all the text into lower case

> Lamatization/stemming
    To find the root word (Remove the tail part of the word)
    eg: amazing > amaze
        studying > study
        leaving > leave 



Tokenization
-------------

eg : " vinay is a good boy " >> split the words using white spaces

>> vinay
   is
   a
   good
   boy 

This word by word output is called tokens This step is done using the n-gram method where n is number of white spaces considered before splitting

n-gram method
-------------

eg: vinay is
    good boy   >> 2-gram method where the words are split after two white spaces

Vectorization
-------------

Converting the tokens into numerical format

TF=Term Frequency
