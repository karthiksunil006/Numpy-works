```python
import pandas as pd
import numpy as np
```

```python
data=pd.read_csv("taxi_trip_pricing.csv")
df=pd.DataFrame(data)
df
```

```python
df.isna().sum()
```

```python
df.dtypes
```

```python
for i in list(df):

    if df[i].dtype=="object":

        df[i]=df[i].fillna(df[i].mode()[0])

    else:
        df[i]=df[i].fillna(df[i].mean())
```

```python
df.isna().sum()
```

```python
df_E=pd.get_dummies(df[["Time_of_Day","Day_of_Week","Traffic_Conditions","Weather"]],dtype=int)
```

```python
df_E.dtypes
```

```python
df_E.info()
```

```python
# combine both df and df_E

final_df=pd.concat([df,df_E],axis=1)
final_df
```

```python
final_df=final_df.drop(columns=["Time_of_Day","Day_of_Week","Traffic_Conditions","Weather"],axis=1)
final_df
```

```python
Y=final_df["Trip_Price"]
```

```python
final_df=final_df.drop(columns=["Trip_Price"],axis=1)
X=final_df
X
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3)
```

```python
from sklearn.preprocessing import StandardScaler

sc=StandardScaler()

X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression

lr=LinearRegression()

lr=lr.fit(X_train,Y_train)

Y_pred=lr.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_percentage_error,r2_score

print(f"Mean_Absolute_Percentag_Error = {mean_absolute_percentage_error(Y_test,Y_pred)}")

print(f"R2_Score = {r2_score(Y_test,Y_pred)}")
```
