```python
import pandas as pd
import numpy as np
```

```python
data=pd.read_csv("Netflix TV Shows and Movies.csv")
df=pd.DataFrame(data)
df
```

```python
df.columns
```

```python
df=df.drop(columns=["index","id","title","description","release_year","imdb_id"],axis=1)
df
```

```python
df.isna().sum()
```

```python
df=df.drop(columns=["age_certification"],axis=1)
df
```

```python
df.isna().sum()
```

```python
df.dtypes
```

```python
df["imdb_votes"]=df["imdb_votes"].fillna(df["imdb_votes"].mean())
```

```python
df.isna().sum()
```

```python
from sklearn.preprocessing import LabelEncoder

le=LabelEncoder()

df["type"]=le.fit_transform(df["type"])
```

```python
Y=df.iloc[:,2]
Y
```

```python
df=df.drop(columns=["imdb_score"],axis=1)
X=df
X
```

```python
from sklearn.model_selection import train_test_split 

X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3)
```

```python
from sklearn.preprocessing import StandardScaler

sc=StandardScaler()

X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression

lr=LinearRegression()

model=lr.fit(X_train,Y_train)
Y_pred=model.predict(X_test)
Y_pred
```

```python
from sklearn.metrics import mean_absolute_percentage_error,r2_score

print(f"mean_absolute_percentage_error = {mean_absolute_percentage_error(Y_test,Y_pred)}")
print(f"R2_Score = {r2_score(Y_test,Y_pred)}")
```
