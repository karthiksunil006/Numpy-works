```python
import pandas as pd
import numpy as np
```

```python
data=pd.read_csv("Test.csv")
df=pd.DataFrame(data)
df
```

```python
# Replace Yes -> 1 and No -> 0  in Ever_Married column
df["Ever_Married"]=df["Ever_Married"].map({"Yes":1,"No":0})
df["Ever_Married"]
```

```python
# Replace Male -> M and Female -> F in gender column

df["Gender"]=df["Gender"].map({"Male":"M","Female":"F"})
df
```

```python
# Replace low,average,high in spending with 1,2,3 respectively

df["Spending_Score"]=df["Spending_Score"].map({"Low":1,"Average":2,"High":3})
df
```

```python
df["Work_Experience"].isna().sum()
df["Work_Experience"]=df["Work_Experience"].fillna(0)
df
```

```python
# replace all value in var_1=cat_6 with category_6

df["Var_1"]=df["Var_1"].replace({"Cat_6":"Category_6"})
df
```

```python
# Replace age values greater than 80 with mean of age value


mean_value=df["Age"].mean()
df["Age"]=df["Age"].apply(lambda a:mean_value if a>80 else a)
df["Age"]
```

```python
# Replace doctor profession with medical

df["Profession"]=df["Profession"].replace({"Doctor":"Medical"})
df
```

```python
# replace segemntation labels (a,b,c,d) with numerical values

df["Segmentation"]=df["Segmentation"].replace({"A":1,"B":2,"C":3,"D":4})
df
```

```python
# Find how many missing values are present in each columns
df.isna().sum()
```

```python
# drop rows where graduated is missing

df=df.dropna(subset=["Graduated"])
df.isna().sum()
```

```python
# Fill the missing values in profession with "Unknown"
df["Profession"]=df["Profession"].fillna("Unknown")
df
```

```python
# remove rows where age is less than 18

df=df.loc[df["Age"]>18,:]
df
```

```python
# identify duplicate rows based on id and remove them

df=df.drop_duplicates(subset=["ID"],keep="first")
df.duplicated(subset=["ID"]).sum()
```

```python
# Display all customers whose spending is "High" (We have converted high to 3 in earlier question)

df.loc[df["Spending_Score"]==3]
```

```python
# Find all married people (Ever_Married=Yes) (we have converted Yes to 1 in earlier question)

df.loc[df["Ever_Married"]==1]
```

```python
# Show customers whose work_Experience is greater than 5 years

df.loc[df["Work_Experience"]>5]
```

```python
# Get customers aged between 30 and 50

df.loc[(df["Age"]>30)&(df["Age"]<50)]
```

```python
# Show customers who are "Doctor" with "High" spending (We have converted "Doctor" to "Medical" and "High" to "3" in earlier questions)

df.loc[(df["Profession"]=="Medical")&(df["Spending_Score"]==3)]
```

```python
# Create a new column "Age_Group" having 
#<30--Young
#30-50--Adult
#>50--Senior

df["Age_Group"]=df["Age"].apply(lambda a:"Young" if a<30 else "Adult" if a>=30 and a<=50 else "Senior")
df
```

```python
# sort the dataset by age (descending) and work_experience (Ascending)


df=df.sort_values(by="Age",ascending=False)
df=df.sort_values(by="Work_Experience",ascending=True)
df
```

```python
# sort the dataset by age (descending) and work_experience (Ascending)

df=df.sort_values(by=["Age","Work_Experience"],ascending=[False,True])
df
```
